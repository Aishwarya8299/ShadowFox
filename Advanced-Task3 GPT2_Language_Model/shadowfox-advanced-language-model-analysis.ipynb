## 6. Experiment 2: Contextual Understanding

In this experiment, GPT-2 is tested using prompts that contain similar words but different contexts. The objective is to observe whether the model can use the surrounding context to generate an appropriate continuation.

The experiment focuses on the ability of GPT-2 to interpret context and distinguish between different meanings of words.

```python
context_prompts = [
    "The bank was crowded because",
    "The river bank was crowded because",
    "The software engineer went to the bank because",
    "The fisherman sat near the river bank because"
]

for i, prompt in enumerate(context_prompts, 1):
    result = generator(
        prompt,
        max_new_tokens=40,
        num_return_sequences=1,
        pad_token_id=generator.tokenizer.eos_token_id
    )
    
    print(f"\n--- Context Prompt {i} ---")
    print("Input:", prompt)
    print("Generated Text:", result[0]["generated_text"])
```

## 6.1 Analysis of Contextual Understanding

The results show that GPT-2 changes its generated text based on the context provided in the prompt.

For example, when the prompt was "The bank was crowded because", the generated text mainly focused on financial and banking-related concepts. However, when the prompt was changed to "The river bank was crowded because", the generated text shifted toward topics related to rivers and environmental situations.

Similarly, the prompt involving a software engineer and a bank produced text related to computer transactions and banking systems. The prompt involving a fisherman near a river bank generated a story related to the fisherman's situation and safety.

These observations suggest that GPT-2 can use surrounding words and context to influence the meaning of ambiguous terms such as "bank".

However, the generated responses also contained information that was not directly supported by the input prompts. For example, some responses introduced topics such as migrants, floods, or other events that were not mentioned in the original input.

Therefore, the experiment suggests that GPT-2 demonstrates contextual text generation capabilities, but its output may not always be logically consistent or factually reliable.

### Key Finding

GPT-2 can generate different responses based on contextual information, but this should not be considered the same as human-level understanding. The model generates text based on patterns learned from its training data and may sometimes produce irrelevant or unsupported information.

## 8. Experiment 4: Effect of Temperature on Text Generation

Temperature is a parameter that controls the randomness of text generation.

A lower temperature generally produces more predictable and conservative text, while a higher temperature can produce more diverse and creative outputs.

In this experiment, the same prompt is provided to GPT-2 using different temperature values. The generated responses are compared to understand how the temperature parameter affects the model's output.

```python
prompt = "The future of artificial intelligence will"

temperatures = [0.3, 0.7, 1.0]

for temperature in temperatures:
    result = generator(
        prompt,
        max_new_tokens=50,
        temperature=temperature,
        do_sample=True,
        num_return_sequences=1,
        pad_token_id=50256
    )

    print(f"\n--- Temperature: {temperature} ---")
    print(result[0]["generated_text"])
```

## 8.1 Analysis of Temperature Experiment

The experiment demonstrates that the temperature parameter influences the diversity and randomness of GPT-2's generated text.

At a lower temperature of 0.3, the generated response was relatively predictable and focused on the general relationship between artificial intelligence and humans.

At a temperature of 0.7, the response became more diverse and included additional concepts such as machine learning, human-computer interaction, and communication.

At a higher temperature of 1.0, the generated response became more varied and introduced new information. However, the response also included unsupported details and was less predictable.

### Key Finding

The experiment suggests that temperature can influence the creativity and diversity of generated text. Lower temperature values tend to produce more predictable outputs, while higher values can produce more diverse outputs. However, increasing randomness does not guarantee better quality, accuracy, or logical consistency.

Therefore, the appropriate temperature depends on the application. A lower temperature may be useful when consistency is important, while a higher temperature may be useful for creative text generation.

## 9.1 Attention Visualization

GPT-2 is based on the Transformer architecture, which uses self-attention mechanisms to process relationships between tokens.

In this experiment, attention weights from GPT-2 are extracted and visualized using a heatmap. The visualization provides an intuitive representation of how the model distributes attention across different input tokens.

The purpose of this experiment is not to claim that the attention map directly represents the complete reasoning process of the model, but rather to provide a visual exploration of the attention patterns produced by the Transformer architecture.

```python
import torch
import matplotlib.pyplot as plt
from transformers import GPT2Tokenizer, GPT2Model

# Load tokenizer and GPT-2 model with attention outputs
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
attention_model = GPT2Model.from_pretrained("gpt2", output_attentions=True)

# Example input
text = "Artificial intelligence is changing the world because technology is becoming more powerful."

# Tokenize input
inputs = tokenizer(text, return_tensors="pt")

# Run model
with torch.no_grad():
    outputs = attention_model(**inputs)

# Get attention weights
attentions = outputs.attentions

# Select the last layer
last_layer_attention = attentions[-1]

# Select the first attention head
attention_matrix = last_layer_attention[0, 0].numpy()

# Get token names
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

# Plot attention heatmap
plt.figure(figsize=(10, 8))

plt.imshow(attention_matrix, aspect="auto")

plt.xticks(range(len(tokens)), tokens, rotation=90)
plt.yticks(range(len(tokens)), tokens)

plt.xlabel("Tokens being attended to")
plt.ylabel("Current tokens")
plt.title("GPT-2 Self-Attention Heatmap")

plt.colorbar(label="Attention Weight")

plt.tight_layout()
plt.show()
```

## 9.2 Baseline Comparison: GPT-2 vs DistilGPT-2

To provide a simple baseline comparison, GPT-2 is compared with DistilGPT-2 using the same input prompt.

DistilGPT-2 is a smaller distilled version of GPT-2 designed to provide faster and more efficient inference.

The comparison focuses on the generated output and generation time. This experiment helps demonstrate the trade-off between model size and computational efficiency.

```python
import time
from transformers import pipeline

prompt = "The future of artificial intelligence will"

# GPT-2
gpt2_generator = pipeline(
    "text-generation",
    model="gpt2",
    pad_token_id=50256
)

start_time = time.time()

gpt2_result = gpt2_generator(
    prompt,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.7,
    num_return_sequences=1
)

gpt2_time = time.time() - start_time


# DistilGPT-2
distilgpt2_generator = pipeline(
    "text-generation",
    model="distilgpt2",
    pad_token_id=50256
)

start_time = time.time()

distilgpt2_result = distilgpt2_generator(
    prompt,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.7,
    num_return_sequences=1
)

distilgpt2_time = time.time() - start_time


# Display results
print("===== GPT-2 =====")
print(gpt2_result[0]["generated_text"])

print("\nGeneration Time:", round(gpt2_time, 2), "seconds")

print("\n===== DistilGPT-2 =====")
print(distilgpt2_result[0]["generated_text"])

print("\nGeneration Time:", round(distilgpt2_time, 2), "seconds")
```

## 9.3 Measuring Repetition in Generated Text

One limitation observed during the experiments was that GPT-2 may sometimes repeat words or phrases.

To quantify this behavior, repeated bigrams (two-word sequences) are counted in generated text.

The repetition rate is calculated as:

Repetition Rate = Number of Repeated Bigrams / Total Number of Bigrams

A higher repetition rate indicates that the generated text contains more repeated two-word sequences.

```python
from collections import Counter
import re

def calculate_repetition_rate(text):
    # Convert text to lowercase and extract words
    words = re.findall(r'\b\w+\b', text.lower())

    # Create bigrams
    bigrams = list(zip(words, words[1:]))

    if len(bigrams) == 0:
        return 0

    # Count occurrences
    counts = Counter(bigrams)

    # Count repeated bigrams
    repeated_bigrams = sum(
        count - 1
        for count in counts.values()
        if count > 1
    )

    repetition_rate = repeated_bigrams / len(bigrams)

    return repetition_rate


# Test on GPT-2 generated text
sample_text = gpt2_result[0]["generated_text"]

repetition_rate = calculate_repetition_rate(sample_text)

print("Generated Text:")
print(sample_text)

print("\nRepetition Rate:",
      round(repetition_rate * 100, 2), "%")
```

```python
# Calculate repetition rates
gpt2_repetition = calculate_repetition_rate(
    gpt2_result[0]["generated_text"]
)

distilgpt2_repetition = calculate_repetition_rate(
    distilgpt2_result[0]["generated_text"]
)

models = ["GPT-2", "DistilGPT-2"]

repetition_rates = [
    gpt2_repetition * 100,
    distilgpt2_repetition * 100
]

plt.figure(figsize=(7, 5))

plt.bar(models, repetition_rates)

plt.xlabel("Model")
plt.ylabel("Repetition Rate (%)")
plt.title("Repetition Rate Comparison")

plt.show()
```
